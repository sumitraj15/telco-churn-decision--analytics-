```python
import pandas as pd

df = pd.read_csv("Telco Churn Data.csv")

print(df.shape)
print(df.info())
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
churn = df["Churn"].value_counts()
print(churn) 
```

```python
churn_prop = round(df["Churn"].value_counts(normalize=True),2)
print(churn_prop)
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors = 'coerce')

print(df["TotalCharges"].isnull().sum())
```

```python
print(df[df["TotalCharges"].isnull()][['tenure','MonthlyCharges','TotalCharges']])
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
print(df["TotalCharges"].isnull().sum())
```

```python
print(df[df["TotalCharges"] == 0][['tenure','MonthlyCharges','TotalCharges']])
```

```python
df["Churn"] = df["Churn"].map({'Yes':1, 'No':0})
```

```python
print(df["SeniorCitizen"].is_unique)

print(df["SeniorCitizen"].value_counts())
print(round(df["SeniorCitizen"].value_counts(normalize=True),2))
```
